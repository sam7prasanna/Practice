```python
def binary_search(lst, a):
    low = 0
    high = len(lst) - 1

    while low<=high:
        mid = (low+high)//2
        if lst[mid] == a:
            return mid
        elif lst[mid] < a:
            low = mid + 1
        else:
            high = mid - 1
    return "Value not in the list"

lst = [2, 5, 9, 13, 16, 19]

print(binary_search(lst, 3))
```

```python
def gcd(a,b):
    while b!=0:
        a,b = b,a%b
    return a

print(gcd(48,18))
```

```python
import datetime
import time
```

```python
ser_list = list(range(0, 10000))
```

```python
# linear search
def lin_ser(ser_lst, a):
    for i in ser_lst:
        if i == a:
            return True
    return False

start = time.time()
lin_ser(ser_list, 9999)
end = time.time()

print(end - start)
```

```python
# binary search
def bin_search(ser_list, a):
    low = 0
    high = len(ser_list) - 1
    
    while low<=high:
        mid = (low+high)//2
        if ser_list[mid] == a:
            return mid
        elif ser_list[mid] < a:
            low = mid + 1
        else:
            high = mid - 1
    return "Value Not Found"

start = time.time()
bin_search(ser_list, 9999)
end = time.time()

print(end - start)
```

```python
# Recursion function

def factorial(n):
    if n == 1:  # base case
        return 1
    else:       # recursive case
        return n * factorial(n-1)
    
print(factorial(2))
```

```python
# Simplified recursion

def fac(n):
    if n==1 or n==0:
        return 1
    return n * fac(n-1)

print(fac(5))
```

```python
# fibonacci series

def fib(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)

print(fib(5))
```

```python
def fib1(n, memo={}):
    if n in memo:
        return memo[n]
    if n == 0:
        return 0
    if n == 1:
        return 1
    memo[n] = fib1(n-1, memo) + fib1(n-2, memo)
    return memo[n]

fib1(6)
```

```python
def countdown(b):
    if b == 0:
        print("Countdown over")
    else:
        print(b)
        return countdown(b-1)

countdown(5)
```

```python
class Car:
    def __init__(self, model, color):
       self.model = model
       self.color = color
    def show(self):
       print(f"Model is {self.model}")
       print(f"Color is {self.color}")
    def start(self):
       print("start & drive the car")
    def stop(self):
       print("Stop the car")
    
# Creating instances of the Car class

audi = Car("audi a4", "blue")
ferrari = Car("ferrari 488", "green")
audi.show()
ferrari.show()

audi.start()
```

```python
class Stack:
    def __init__(self):
        self.items = []
    
    def push(self, item):
        self.items.append(item)
    
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        else:
            return 'Stack is empty'
        
    def is_empty(self):
        return len(self.items) == 0
    
s = Stack()

print(s.push(10))
print(s.push(20))
# s.pop

print(s.pop())   # 20
print(s.pop())   # 10
print(s.pop())   # Stack is empty
```

```python
stack = []

# append() function to push element in the stack
stack.append('g')
stack.append('f')
stack.append('g')

print('Initial stack')
print(stack)

# pop() function to pop element from stack in LIFO order
print('\nElements popped from stack:')
print(stack.pop())
print(stack.pop())
print(stack.pop())

print('\nStack after elements are popped:')
print(stack)

# uncommenting print(stack.pop()) will cause an IndexError as the stack is now empty
```
