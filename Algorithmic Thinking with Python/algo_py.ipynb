```python
def binary_search(lst, a):
    low = 0
    high = len(lst) - 1

    while low<=high:
        mid = (low+high)//2
        if lst[mid] == a:
            return mid
        elif lst[mid] < a:
            low = mid + 1
        else:
            high = mid - 1
    return "Value not in the list"

lst = [2, 5, 9, 13, 16, 19]

print(binary_search(lst, 3))
```

```python
def gcd(a,b):
    while b!=0:
        a,b = b,a%b
    return a

print(gcd(48,18))
```

```python
import datetime
import time
```

```python
ser_list = list(range(0, 10000))
```

```python
# linear search
def lin_ser(ser_lst, a):
    for i in ser_lst:
        if i == a:
            return True
    return False

start = time.time()
lin_ser(ser_list, 9999)
end = time.time()

print(end - start)
```

```python
# binary search
def bin_search(ser_list, a):
    low = 0
    high = len(ser_list) - 1
    
    while low<=high:
        mid = (low+high)//2
        if ser_list[mid] == a:
            return mid
        elif ser_list[mid] < a:
            low = mid + 1
        else:
            high = mid - 1
    return "Value Not Found"

start = time.time()
bin_search(ser_list, 9999)
end = time.time()

print(end - start)
```
