using while

```python
for i in range(1, 10):
    print(i)
```

```python
i = 1

while(i <= 10):
    print(i)
    i += 1
```

```python
def printNos(n):
    if n == 0:
        
        # base condition
        return
    
    # recursive call first
    printNos(n - 1)
  
    # print after recursion
    print(n, end=' ')


if __name__ == "__main__":
    n = 10
    printNos(n)
```

2️. Sum of numbers from 1 to N.  Take a number N and find the sum from 1 to N.

using for-if

using for

using for(start point, end point)

using while

using math formula

```python
n = 10
count = 0

for i in range(n+1):
    if i <= n:
        count = count + i

print(count)
```

```python
n = 10
count = 0

for i in range(n+1):
    count = count + i

print(count)
```

```python
count = 0

for i in range(1, n+1):
    count = count + i

print(count)
```

```python
count = 0
i = 0

while(i <= 10):
    count = count + i
    i = i + 1

print(count)
```
